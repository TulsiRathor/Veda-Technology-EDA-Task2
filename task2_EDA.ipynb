```python
plt.figure(figsize=(8, 5))

plt.hist(df['Sales'], bins=30)

plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")

plt.show()
```

### Insight : Sales Distribution

The histogram shows that most sales values are concentrated in the lower range, while only a few transactions have very high sales values. The distribution is positively skewed (right-skewed), indicating that high-value sales are relatively less frequent.

```python
plt.figure(figsize=(8, 5))

plt.hist(df['Profit'], bins=30)

plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x = df['Sales'], color = 'orange')
plt.title('Box Plot of Sales (Outlier Detection)')
plt.show()
```

### Insight: Box Plot Of Sales(Outlier Detection)
* **Concentration of Typical Sales:** The main box (interquartile range) is tightly compressed near the left side(close to 0) ,Showing that the vast majority  of sales orders of relatively small dollar amounts.
* **Heavy Presence Of Outliers :** there is long trail of individual data points extending from around 1000 up to over 22000,indicating a large number of high-value outliers.
* **Skewness :** The data is heavily right - skewed.

```python
plt.figure(figsize=(8, 5))
sns.barplot(x = df['Category'] , y =df['Sales'])
plt.title('Total Sales by Category')
# plt.xticks(rotation = 45)
plt.show()
```

### Insight : Sales by Category
* **Top Performing Category :** **Technology** leads all categories with highest average sales (around 440).
* **Second Highest :** **Furniture** ranks second (around 440),contributing a solid mid-to-high tier of sales performance.
* **Lowest Sales Value :** **Office Supplies** has the lowest sales figure per transaction (around 120),likely because these items consist of smaller,everyday essential sold at lower unit prices.

```python
plt.figure(figsize = (8, 6))
numeric_df = df.select_dtypes(include=[np.number])
sns.heatmap(numeric_df.corr(),annot = True)
plt.title('Corelation Heatmap')
plt.show()
                              
```

### Insight : Correlation Heatmap
* **Strongest Positive Relationship(Sales vs. Profit = 0.48):** There is moderate positive correlation of **0.48** between **sales** and **Profit**,indicating that higher sales transactions generally lead to higher profits.
* **Negative Impact of Discounts(Discount vs. Profit = -0.22) :** **Discount** and **Profit** have a negative correlation of **-0.22**.showing that offering higher discounts reduces overall profitability.
* **Weak/No Correlation With Quantity & Row ID:** **Quantity** has only a weak positive correlation with **Sales(0.20)** and almost no effect on **Profit (0.066)**.**Row ID** shows values near zero across all variables,confirming it is simply an index identifier with no analytical impact.

```python
catg_sales = df.groupby('Category')['Sales'].sum()
plt.figure(figsize = (8,4))
plt.plot(catg_sales.index,catg_sales.values, marker ='o' , color = 'blue' ,linewidth =2)
plt.title('Total Sales by Category')
plt.xlabel('Category')
plt.ylabel('Total Sales')
plt.grid(True)
plt.show()
```

### insight : Total Sales by Category (Line Chart)
* **Highest Revenue Driver:** **Technology** achievs the highest total sales (over 830,000), showing  a sharp upward jump compared to the other two categories .
* **Steady Mid-Tier:** **Furniture** holds the second position with total sales around 740,000.
* **Lowest Total Sales:** **Office Supplies** records the lowest cumulative sales (around 719,000) ,showing a slight dip between furniture and technology .
